```python
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/iris-dataset.csv")
df.head()
```

```python
df.hist(figsize=(10,6))
plt.suptitle("Distribución de variables numéricas - Dataset limpio")
plt.show()
```

```python
df.groupby("Species").mean()
```

```python
df.columns
```

```python
df.boxplot(by="Species", figsize=(10, 6))
plt.suptitle("Comparación de variables por especie")
plt.title("")
plt.show()
```

```python
df.hist(figsize=(10,6))
plt.suptitle("Distribución de vatiables - Después de limpieza")
plt.savefig("../reports/figures/histogram_cleaned_data.png")
plt.close()
```

## Conclusiones

- El dataset presenta distribuciones coherentes tras el proceso de limpieza.
- Se observan diferencias claras entre especies en las variables analizadas.
- El flujo ETL permitió obtener un dataset listo para análisis.
